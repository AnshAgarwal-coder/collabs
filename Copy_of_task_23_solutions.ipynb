### **`Problem-1:`** Draw a line plot of which, the x-axis is the "Year" and the y-axis is sum of "PM2.5" of two countries Iran and China.

```python
df= pd.read_csv('https://tinyurl.com/2fe6vz4u')
df.head()
```

```python
iran = df.query('Country == "Iran"').groupby('Year')['PM2.5'].sum()
china = df.query('Country == "China"').groupby('Year')['PM2.5'].sum()
```

```python
plt.plot(iran.index,iran.values,label='Iran',linestyle='dashed')
plt.plot(china.index,china.values,label='China',linestyle='dashed')
plt.title('china and iran AQI')
plt.xlabel('Year')
plt.ylabel('PM2.5 level')
plt.legend(loc = 'best')
plt.xticks(df['Year'].value_counts().index) #x axis label proper formatting
plt.grid()
plt.show()
```

###`Problem-9` Show Quarter wise Sales data for all products as multi Bar chart.


```python
df['date'] = pd.to_datetime(['2026-{}-01'.format(month) for month in df['month_number']])
df.head()
```

```python
df['qua']=df['date'].dt.quarter
final_df = df.groupby(df['qua'])[['facecream','facewash','toothpaste','bathingsoap','shampoo','moisturizer']].sum()
```

```python
i=-1
for col in final_df:
  plt.bar(final_df.index+i,final_df[col],width=0.2,label=col)
  i = i-0.15

plt.xticks(final_df.index - 1.4,final_df.index)
plt.xlabel('Quarter')
plt.ylabel('Sales')
plt.title('Quarter-wise Sales for All Products')
plt.legend(loc='best')
plt.grid(axis='y')
plt.show()
```

```python
#gpt way

df['qua'] = df['date'].dt.quarter

final_df = df.groupby(df['qua'])[['facecream','facewash','toothpaste','bathingsoap','shampoo','moisturizer']].sum()

# positions of the 4 quarters
x = np.arange(len(final_df))

# 6 products → 6 bars around each quarter
width = 0.15
offsets = [-0.375, -0.225, -0.075, 0.075, 0.225, 0.375]

for col, offset in zip(final_df.columns, offsets):
    plt.bar(x + offset, final_df[col], width=width, label=col)

plt.xticks(x, final_df.index)
plt.xlabel('Quarter')
plt.ylabel('Sales')
plt.title('Quarter-wise Sales for All Products')
plt.legend(loc='best')
plt.grid(axis='y')
plt.show()
```

###`Problem-10` Plot Stacked Bar chart quarter wise for each product.

```python
final_df
all_col=[]
for col in final_df:
  if all_col == 0:
    plt.bar(final_df.index,final_df[col],label=col)
  else:
    plt.bar(final_df.index,final_df[col],label=col,bottom=sum(all_col))
    all_col.append(final_df[col])
plt.xticks(final_df.index - 0.02, final_df.index)
plt.legend(loc='best')
plt.show()
```
